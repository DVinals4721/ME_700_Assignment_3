```python
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import HTML, display, Image
from finiteelementanalysis import pre_process as pre
from finiteelementanalysis import pre_process_demo_helper_fcns as pre_demo
from finiteelementanalysis.solver import hyperelastic_solver
from finiteelementanalysis import visualize as viz

# Enable inline plotting for Jupyter Notebook
%matplotlib inline

print("Cantilever Beam FEA: h-refinement and p-refinement Study")

# Beam geometry
L = 10.0   # length in x
H = 1.0    # height in y

# Material properties
E = 1000.0
nu = 0.3
mu = E / (2.0 * (1.0 + nu))
kappa = E / (3.0 * (1.0 - 2.0 * nu))
material_props = np.array([mu, kappa])

print(f"Material properties: mu={mu:.3f}, kappa={kappa:.3f}")

# Load (large load for nonlinear behavior)
q = -0.5  # load per unit length in x

# Solver settings
nr_num_steps = 20
nr_tol = 1e-10
nr_maxit = 30

# Analytical solution (note: this will not be accurate for large deformations)
E_eff = E / (1 - nu ** 2.0)
I = H ** 3 / 12.0
w_analytical = q * L ** 4 / (8.0 * E_eff * I)

def run_analysis(nx, ny, ele_type):
    coords, connect = pre.generate_rect_mesh_2d(ele_type, 0.0, 0.0, L, H, nx, ny)
    
    boundary_nodes, boundary_edges = pre.identify_rect_boundaries(
        coords, connect, ele_type, x_lower=0.0, x_upper=L, y_lower=0.0, y_upper=H
    )
    
    fixed_left = pre.assign_fixed_nodes_rect(boundary_nodes, "left", 0.0, 0.0)
    dload_info = pre.assign_uniform_load_rect(boundary_edges, "top", 0.0, q)
    fixed_nodes = fixed_left
    
    displacements_all, nr_info_all = hyperelastic_solver(
        material_props, ele_type, coords.T, connect.T, fixed_nodes, dload_info,
        nr_print=True, nr_num_steps=nr_num_steps, nr_tol=nr_tol, nr_maxit=nr_maxit,
    )
    
    final_disp = displacements_all[-1]
    
    tip_node = None
    tol = 1e-3
    for i, (x, y) in enumerate(coords):
        if abs(x - L) < tol and abs(y - H/2) < H/(2*ny):
            tip_node = i
            break
    if tip_node is None:
        raise ValueError("Could not find tip node near x=L, y=H/2.")
    
    tip_disp_y = final_disp[2*tip_node + 1]
    
    return tip_disp_y, coords, connect, displacements_all

print("\nh-refinement study")
h_refinement_factors = [2, 5, 10, 15, 20]
h_refinement_results = []
h_refinement_errors = []

for n in h_refinement_factors:
    print(f"\nAnalyzing mesh with {n} elements in x-direction")
    tip_disp, coords, connect, displacements = run_analysis(n, 2, "D2_nn6_tri")
    h_refinement_results.append(tip_disp)
    h_refinement_errors.append(abs(tip_disp - w_analytical))
    
    # Create element quality histograms
    mesh_name = f"Beam_h_{n}"
    aspect_ratios, skewness, min_angles, max_angles = pre_demo.compute_element_quality_metrics("D2_nn6_tri", coords, connect)
    cond_nums, jac_dets = pre_demo.compute_condition_and_jacobian("D2_nn6_tri", coords, connect)
    plot_file = mesh_name + "_histograms"
    pre_demo.plot_element_quality_histograms(
        fname=str(plot_file),
        super_title=f"Mesh Quality Metrics (h-refinement, n={n})",
        ele_type="D2_nn6_tri",
        cond_nums=cond_nums,
        jac_dets=jac_dets,
        aspect_ratios=aspect_ratios,
        skewness=skewness,
        min_angles=min_angles,
        max_angles=max_angles
    )
    display(Image(filename=f"{plot_file}.png"))
    
    # Create deformation GIF
    img_name = f"h_refinement_{n}.gif"
    viz.make_deformation_gif(displacements, coords, connect, "D2_nn6_tri", img_name)
    display(HTML(f'<img src="{img_name}" width="600" loop>'))

# Plot h-refinement results
plt.figure(figsize=(10, 6))
plt.plot(h_refinement_factors, h_refinement_results, marker='o')
plt.axhline(y=w_analytical, color='r', linestyle='--', label='Analytical solution')
plt.xlabel("Number of elements in x-direction")
plt.ylabel("Tip displacement")
plt.title("h-refinement study")
plt.legend()
plt.grid(True)
plt.show()

# Plot h-refinement errors
plt.figure(figsize=(10, 6))
plt.plot(h_refinement_factors, h_refinement_errors, marker='o')
plt.xlabel("Number of elements in x-direction")
plt.ylabel("Absolute error")
plt.title("h-refinement error")
plt.yscale('log')
plt.grid(True)
plt.show()

print("\np-refinement study")
element_types = ["D2_nn3_tri", "D2_nn4_quad", "D2_nn6_tri", "D2_nn8_quad"]
p_refinement_results = []
p_refinement_errors = []

for ele_type in element_types:
    print(f"\nAnalyzing element type: {ele_type}")
    tip_disp, coords, connect, displacements = run_analysis(20, 2, ele_type)
    p_refinement_results.append(tip_disp)
    p_refinement_errors.append(abs(tip_disp - w_analytical))
    
    # Create element quality histograms
    mesh_name = f"Beam_p_{ele_type}"
    aspect_ratios, skewness, min_angles, max_angles = pre_demo.compute_element_quality_metrics(ele_type, coords, connect)
    cond_nums, jac_dets = pre_demo.compute_condition_and_jacobian(ele_type, coords, connect)
    plot_file = mesh_name + "_histograms"
    pre_demo.plot_element_quality_histograms(
        fname=str(plot_file),
        super_title=f"Mesh Quality Metrics (p-refinement, {ele_type})",
        ele_type=ele_type,
        cond_nums=cond_nums,
        jac_dets=jac_dets,
        aspect_ratios=aspect_ratios,
        skewness=skewness,
        min_angles=min_angles,
        max_angles=max_angles
    )
    display(Image(filename=f"{plot_file}.png"))
    
    # Create deformation GIF
    img_name = f"p_refinement_{ele_type}.gif"
    viz.make_deformation_gif(displacements, coords, connect, ele_type, img_name)
    display(HTML(f'<img src="{img_name}" width="600" loop>'))

# Plot p-refinement results
plt.figure(figsize=(10, 6))
plt.plot(element_types, p_refinement_results, marker='o')
plt.axhline(y=w_analytical, color='r', linestyle='--', label='Analytical solution')
plt.xlabel("Element type")
plt.ylabel("Tip displacement")
plt.title("p-refinement study")
plt.legend()
plt.grid(True)
plt.show()

# Plot p-refinement errors
plt.figure(figsize=(10, 6))
plt.plot(element_types, p_refinement_errors, marker='o')
plt.xlabel("Element type")
plt.ylabel("Absolute error")
plt.title("p-refinement error")
plt.yscale('log')
plt.grid(True)
plt.show()

# Display results
print(f"\nAnalytical solution: {w_analytical:.6f}")

print("\nh-refinement results:")
for n, result, error in zip(h_refinement_factors, h_refinement_results, h_refinement_errors):
    print(f"nx = {n}: tip displacement = {result:.6f}, error = {error:.6e}")

print("\np-refinement results:")
for ele_type, result, error in zip(element_types, p_refinement_results, p_refinement_errors):
    print(f"{ele_type}: tip displacement = {result:.6f}, error = {error:.6e}")
```
